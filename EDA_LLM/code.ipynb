```python
import ollama   # ✅ CORRECT
def generate_insights(df_summary):
     prompt = f"Analyze the dataset summary and provide insights:\\n\\n{df_summary}"
     response = ollama.chat(model="mistral", messages=[{"role": "user", "content": prompt}])
     return response['message']['content']
# Generate AI Insights\n,
summary = df.describe().to_string()
insights = generate_insights(summary)
print("\n🔹 AI-Generated Insights:\n", insights)
```

```python
import gradio as gr
import pandas as pd

# Dummy insights function (replace with your AI model)
def generate_insights(summary):
    return f"EDA Summary:\n\n{summary}"

def eda_analysis(file):
    try:
        df = pd.read_csv(file.name)
        summary = df.describe().to_string()
        insights = generate_insights(summary)
        return insights
    except Exception as e:
        return f"Error: {str(e)}"

# Create Web Interface (OUTSIDE function)
demo = gr.Interface(
    fn=eda_analysis,
    inputs=gr.File(),
    outputs="text",
    title="AI-Powered EDA with Mistral"
)

# Launch App
demo.launch(share=True)
```
